```python
import mne

COMMON_CHS = [
    "Fp1", "Fp2",
    "F7", "F3", "Fz", "F4", "F8",
    "C3", "Cz", "C4",
    "P3", "Pz", "P4",
    "O1", "O2",
]

def preprocess_for_transfer(raw, sfreq=250):
    raw = raw.copy()

    # Keep identical sensor space across datasets
    raw.pick(COMMON_CHS)

    # Standard electrode coordinates
    montage = mne.channels.make_standard_montage("standard_1020")
    raw.set_montage(montage, on_missing="raise")

    # Same sampling rate
    raw.resample(sfreq)

    # Same filtering
    raw.filter(1.0, 45.0)
    raw.notch_filter(60.0)

    # Same reference, computed over exactly the same 15 electrodes
    raw.set_eeg_reference(ref_channels="average", projection=False)

    return raw
```

```python
raw_alz = mne.io.read_raw_eeglab(alz_path, preload=True)
raw_pearl = mne.io.read_raw_brainvision(pearl_path, preload=True)

raw_alz = preprocess_for_transfer(raw_alz)
raw_pearl = preprocess_for_transfer(raw_pearl)
```

```python
from pathlib import Path
import numpy as np

def load_npz_dataset(processed_dir):
    processed_dir = Path(processed_dir)

    xs = []
    labels = []
    groups = []
    subject_ids = []
    ages = []
    sex_or_gender = []

    for path in sorted(processed_dir.glob("*.npz")):
        data = np.load(path, allow_pickle=True)

        xs.append(data["x"])
        labels.append(int(data["label"]))
        groups.append(str(data["group"]))
        subject_ids.append(str(data["participant_id"]))
        ages.append(int(data["age"]))

        if "sex" in data.files:
            sex_or_gender.append(int(data["sex"]))
        else:
            sex_or_gender.append(str(data["gender"]))

    x = np.concatenate(xs, axis=0)

    epoch_labels = np.concatenate([
        np.full(len(subject_x), label)
        for subject_x, label in zip(xs, labels)
    ])

    epoch_subject_ids = np.concatenate([
        np.full(len(subject_x), subject_id)
        for subject_x, subject_id in zip(xs, subject_ids)
    ])

    return {
        "x": x,
        "epoch_labels": epoch_labels,
        "epoch_subject_ids": epoch_subject_ids,
        "subject_ids": np.array(subject_ids),
        "subject_labels": np.array(labels),
        "subject_groups": np.array(groups),
        "ages": np.array(ages),
        "sex_or_gender": np.array(sex_or_gender),
    }
```

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

df = pd.read_csv(
    "/data/s.dharia-ra/difflogic_AD_PEARL/outputs/difflogic/medium/summary/pearl_subject_predictions_seed_average.csv"
)

observed = (
    df.loc[df["group"] == "A+P+", "mean_p_ad"].median()
    - df.loc[df["group"] == "N", "mean_p_ad"].median()
)

n_perm = 100000
null_stats = []

groups = df["group"].to_numpy()
scores = df["mean_p_ad"].to_numpy()

for _ in range(n_perm):
    shuffled_groups = rng.permutation(groups)

    stat = (
        np.median(scores[shuffled_groups == "A+P+"])
        - np.median(scores[shuffled_groups == "N"])
    )
    null_stats.append(stat)

null_stats = np.array(null_stats)

p_perm = (np.sum(null_stats >= observed) + 1) / (n_perm + 1)

print("Observed A+P+ - N median difference:", observed)
print("Permutation p-value:", p_perm)
```

```python
import numpy as np
import pandas as pd
from scipy.stats import kruskal

rng = np.random.default_rng(42)

df = pd.read_csv(
    "/data/s.dharia-ra/difflogic_AD_PEARL/outputs/difflogic/medium/summary/pearl_subject_predictions_seed_average.csv"
)

scores = df["mean_p_ad"].to_numpy()
groups = df["group"].to_numpy()

def median_diff(df, group_a, group_b):
    return (
        df.loc[df["group"] == group_a, "mean_p_ad"].median()
        - df.loc[df["group"] == group_b, "mean_p_ad"].median()
    )

def permutation_median_diff(df, group_a, group_b, n_perm=10000, seed=42):
    rng = np.random.default_rng(seed)

    observed = median_diff(df, group_a, group_b)

    scores = df["mean_p_ad"].to_numpy()
    groups = df["group"].to_numpy()

    null_stats = []

    for _ in range(n_perm):
        shuffled_groups = rng.permutation(groups)

        stat = (
            np.median(scores[shuffled_groups == group_a])
            - np.median(scores[shuffled_groups == group_b])
        )
        null_stats.append(stat)

    null_stats = np.array(null_stats)

    p_one_sided = (np.sum(null_stats >= observed) + 1) / (n_perm + 1)
    p_two_sided = (np.sum(np.abs(null_stats) >= abs(observed)) + 1) / (n_perm + 1)

    return observed, p_one_sided, p_two_sided

comparisons = [
    ("A+P+", "N"),
    ("A+P+", "A+P-"),
    ("A+P-", "N"),
]

for group_a, group_b in comparisons:
    observed, p_one, p_two = permutation_median_diff(df, group_a, group_b)

    print(f"{group_a} - {group_b}")
    print(f"Observed median difference: {observed:.4f}")
    print(f"Permutation p one-sided:  {p_one:.4f}")
    print(f"Permutation p two-sided:  {p_two:.4f}")
    print()
```

```python
def kruskal_statistic(scores, groups):
    values = [
        scores[groups == group]
        for group in ["N", "A+P-", "A+P+"]
    ]
    return kruskal(*values).statistic

def permutation_kruskal(df, n_perm=10000, seed=42):
    rng = np.random.default_rng(seed)

    scores = df["mean_p_ad"].to_numpy()
    groups = df["group"].to_numpy()

    observed = kruskal_statistic(scores, groups)

    null_stats = []

    for _ in range(n_perm):
        shuffled_groups = rng.permutation(groups)
        null_stats.append(kruskal_statistic(scores, shuffled_groups))

    null_stats = np.array(null_stats)

    p_perm = (np.sum(null_stats >= observed) + 1) / (n_perm + 1)

    return observed, p_perm

observed_h, p_perm = permutation_kruskal(df)

print("Overall 3-group Kruskal permutation")
print(f"Observed H statistic: {observed_h:.4f}")
print(f"Permutation p-value: {p_perm:.4f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

base = "/data/s.dharia-ra/difflogic_AD_PEARL"

pearl = pd.read_csv(
    f"{base}/outputs/difflogic/medium/summary/pearl_subject_predictions_seed_average.csv"
)

rng = np.random.default_rng(42)

scores = pearl["mean_p_ad"].to_numpy()
groups = pearl["group"].to_numpy()

observed = (
    pearl.loc[pearl["group"] == "A+P+", "mean_p_ad"].median()
    - pearl.loc[pearl["group"] == "N", "mean_p_ad"].median()
)

null_stats = []

for _ in range(10000):
    shuffled = rng.permutation(groups)
    stat = (
        np.median(scores[shuffled == "A+P+"])
        - np.median(scores[shuffled == "N"])
    )
    null_stats.append(stat)

plt.figure(figsize=(6, 4))
plt.hist(null_stats, bins=40, color="lightgray", edgecolor="black")
plt.axvline(observed, color="red", linewidth=2, label=f"Observed = {observed:.3f}")
plt.xlabel("Permuted median difference: A+P+ - N")
plt.ylabel("Count")
plt.title("Permutation null: A+P+ vs N")
plt.legend()
plt.tight_layout()
plt.show()
```
